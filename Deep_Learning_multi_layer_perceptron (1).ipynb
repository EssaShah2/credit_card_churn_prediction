```python
import pandas as pd
```

```python
df = pd.read_csv("Customer-Churn-Records.csv")
df.head()
```

```python
df.drop(columns=['RowNumber','CustomerId','Surname'], inplace=True)
df.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x = df.drop(columns=['Exited'])
y = df['Exited']
x_train, x_test , y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
trf1 = ColumnTransformer([
    ('onehot',OneHotEncoder(drop='first', sparse_output=False , handle_unknown= 'ignore'),['Geography','Gender','Card Type'])
],remainder='passthrough')
```

```python
trf1.set_output(transform='pandas')
x_train = trf1.fit_transform(x_train)
x_test = trf1.transform(x_test)
```

```python
x_train
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
x_train = pd.DataFrame(x_train, columns=trf1.get_feature_names_out())
x_test = pd.DataFrame(x_test, columns=trf1.get_feature_names_out())
```

```python
x_train
```

```python
from imblearn.over_sampling import SMOTE
smote = SMOTE()
x_train,y_train = smote.fit_resample(x_train,y_train)
x_test , y_test = smote.fit_resample(x_test , y_test)
```

```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
x_train.shape
```

```python
model = Sequential()
model.add(Dense(5, activation = 'sigmoid', input_dim = 17))
model.add(Dense(5, activation = 'sigmoid'))
model.add(Dense(1, activation = 'sigmoid'))
```

```python
model.compile(optimizer = 'Adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
```

```python
model.fit(x_train, y_train,epochs =  10, validation_split= 0.2)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report
import numpy as np
y_pred = np.where(y_pred>0.5,1,0)
```

```python
accuracy_score(y_test ,y_pred)
```

```python
print(classification_report(y_test , y_pred))
```
